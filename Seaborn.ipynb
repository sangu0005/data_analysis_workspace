## countplot()

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Hr_Data.csv')
```

```python
df
```

```python
df['department'].unique()
```

```python
plt.figure(figsize=(12,5))
order = ['hr', 'accounting', 'IT', 'product_mng', 'marketing', 'RandD', 'management', 'technical', 'support' , 'sales']
sns.countplot(data=df, x='department', hue='salary', order=order, hue_order=['high', 'medium', 'low'], palette='colorblind')
plt.gca().set_facecolor('lightgrey')
plt.gcf().set_facecolor('lightyellow')
```

```python
plt.figure(figsize=(12,5))
order = ['hr', 'accounting', 'IT', 'product_mng', 'marketing', 'RandD', 'management', 'technical', 'support' , 'sales']
sns.countplot(data=df, y='salary', hue='department', order=['high', 'medium', 'low'], hue_order=order, palette='husl')
plt.gca().set_facecolor('lightyellow')
plt.gcf().set_facecolor('lightblue')
```
