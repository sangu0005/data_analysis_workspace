## countplot()

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Hr_Data.csv')
```

```python
df
```

```python
df['department'].unique()
```

```python
plt.figure(figsize=(12,5))
order = ['hr', 'accounting', 'IT', 'product_mng', 'marketing', 'RandD', 'management', 'technical', 'support' , 'sales']
sns.countplot(data=df, x='department', hue='salary', order=order, hue_order=['high', 'medium', 'low'], palette='colorblind')
plt.gca().set_facecolor('lightgrey')
plt.gcf().set_facecolor('lightyellow')
```

```python
plt.figure(figsize=(12,5))
order = ['hr', 'accounting', 'IT', 'product_mng', 'marketing', 'RandD', 'management', 'technical', 'support' , 'sales']
sns.countplot(data=df, y='salary', hue='department', order=['high', 'medium', 'low'], hue_order=order, palette='husl')
plt.gca().set_facecolor('lightyellow')
plt.gcf().set_facecolor('lightblue')
```

# Barplot

```python
plt.figure(figsize=(10,6))
ax = sns.barplot(data=df, x='department',y='number_project',palette='Set1')
plt.gca().set_facecolor('lightgrey')
plt.gcf().set_facecolor('lightyellow')

for container in ax.containers:
    ax.bar_label(container, fmt='%.2f',padding=3)

plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(data=df, x='department',y='number_project',hue='salary',palette='colorblind',estimator=np.max)
plt.gca().set_facecolor('lightgrey')
plt.gcf().set_facecolor('lightyellow')
plt.legend(bbox_to_anchor=(1.2,1))
plt.show()
```
