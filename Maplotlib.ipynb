```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(1,8)
y = np.arange(9,16)
plt.plot(x,y)
plt.grid()
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Plot')
```

```python
z = x * y
plt.plot(x,y)
plt.plot(x,z)
```

```python
plt.plot(x,y, label = 'X vs Y')
plt.plot(x,z, label = 'X vs Z')
plt.plot(x, x**2, label = 'X vs X^2')
plt.grid(axis='both')
plt.legend(loc='upper left')
```
