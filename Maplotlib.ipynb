```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(1,8)
y = np.arange(9,16)
plt.plot(x,y)
plt.grid()
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Plot')
```

```python
z = x * y
plt.plot(x,y)
plt.plot(x,z)
```

```python
plt.plot(x,y, label = 'X vs Y')
plt.plot(x,z, label = 'X vs Z')
plt.plot(x, x**2, label = 'X vs X^2')
plt.grid(axis='both')
plt.legend(loc='upper left')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Multiple Plots with Legend')
```

```python
plt.subplot(2,2,1)
plt.plot(x,y, label = 'X vs Y',color = 'blue', marker = 'o', markersize = 10, mfc = 'red', mec = 'black')
plt.grid(ls= 'dotted', color = 'green')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Subplot 1')
plt.legend()
plt.gca().set_facecolor('lightpink')

plt.subplot(2,2,2)
plt.plot(x,y, label = 'X vs Y',color = 'blue', marker = 'o', markersize = 10, mfc = 'green', mec = 'black')
plt.grid(ls= 'dotted', color = 'green')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Subplot 2')
plt.legend()
plt.gca().set_facecolor('lightgreen')

plt.subplot(2,2,3)
plt.plot(x,y, label = 'X vs Y',color = 'blue', marker = 'o', markersize = 10, mfc = 'pink', mec = 'black')
plt.grid(ls= 'dotted', color = 'green')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Subplot 3')
plt.legend()
plt.gca().set_facecolor('lightblue')

plt.subplot(2,2,4)
plt.plot(x,y, label = 'X vs Y',color = 'blue', marker = 'o', markersize = 10, mfc = 'violet', mec = 'black')
plt.grid(ls= 'dotted', color = 'green')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Subplot 4')
plt.legend()
plt.gca().set_facecolor('lightyellow')

plt.suptitle('Multiple Subplots Example', fontsize=16)
plt.tight_layout()
plt.gcf().set_facecolor('lightgray')
```

```python
fig, axes = plt.subplots(3, 2)
fig.suptitle('Multiple Subplots with Different Sizes', fontsize=16)
fig.set_facecolor('lightgray')

axes[0, 0].plot(x, y, label='X vs Y', color='blue', marker='o', markersize=10, mfc='red', mec='black')
axes[0, 0].grid(ls='dotted', color='green') 
axes[0, 0].set_xlabel('X-axis')
axes[0, 0].set_ylabel('Y-axis')

axes[0, 1].plot(x, y, label='X vs Y', color='blue', marker='o', markersize=10, mfc='green', mec='black',ls = '-.')
axes[0, 1].grid(ls='dashed', color='green')
axes[0, 1].set_xlabel('X-axis')
axes[0, 1].set_ylabel('Y-axis')
axes[0, 1].set_facecolor('lightgreen')

axes[1, 0].plot(x, y, label='X vs Y', color='blue', marker='o', markersize=10, mfc='pink', mec='black',ls = '--')
axes[1, 0].grid(ls='dotted', color='green')
axes[1, 0].set_xlabel('X-axis')
axes[1, 0].set_ylabel('Y-axis')
axes[1, 0].set_facecolor('lightyellow')

axes[1, 1].plot(x, y, label='X vs Y', color='blue', marker='o', markersize=10, mfc='violet', mec='black',ls = ':')  
axes[1, 1].grid(ls='dotted', color='green')
axes[1, 1].set_xlabel('X-axis')
axes[1, 1].set_ylabel('Y-axis')
axes[1, 1].set_facecolor('lightyellow')

axes[2, 0].plot(x, y, label='X vs Y', color='blue', marker='o', markersize=10, mfc='orange', mec='black',ls = '-')
axes[2, 0].grid(ls='dotted', color='green')
axes[2, 0].set_xlabel('X-axis')
axes[2, 0].set_ylabel('Y-axis')
axes[2, 0].set_facecolor('lightblue')

axes[2, 1].plot(x, y, label='X vs Y', color='blue', marker='o', markersize=10, mfc='cyan', mec='black',ls = '-')
axes[2, 1].grid(ls='dotted', color='green') 
axes[2, 1].set_xlabel('X-axis')
axes[2, 1].set_ylabel('Y-axis')
axes[2, 1].set_facecolor('lightyellow')

fig.tight_layout()
```

```python
marks = [90, 85, 78, 92, 88]
subjects = ['Math', 'Science', 'English', 'History', 'Art']
plt.pie(marks, labels=subjects, autopct='%1.1f%%',shadow=True, startangle=150, colors=['blue', 'green', 'pink', 'red', 'gray'], explode=(0, 0, 0, 0.1, 0))
plt.title('Subject-wise Distribution of Marks')
plt.axis('equal')
plt.legend()
plt.gcf().set_facecolor('lightyellow')
plt.show()
```

```python
points = [16, 14, 12, 10, 8, 6, 4, 2]
teams = ['RCB', 'PBKS', 'MI', 'KKR', 'DC', 'RR', 'SRH', 'CSK']
plt.pie(points, labels=teams, autopct='%1.1f%%',shadow=True, startangle=150, colors=['red', 'green', 'pink', 'blue', 'gray', 'pink', 'orange', 'yellow'], explode=(0.2, 0.1, 0, 0, 0, 0, 0, 0))
plt.pie([1], colors=['white'], radius=0.2)
plt.title('Team-wise Distribution of Points')
```

# Bar chart

```python
score = [565, 688, 565, 953, 678]
teams = ['RR', 'KKR', 'MI', 'CSK', 'LSG']
d = { 'family': 'serif', 'color':  'black', 'weight': 'normal', 'size': 12}
plt.bar(teams, score, color=['red', 'green', 'blue', 'orange', 'purple'], width=0.5, label=teams)
plt.title('Highest Runs of Kohli in IPL', fontdict=d)
plt.xlabel('vs Teams', fontdict=d)
plt.ylabel('Runs Scored', fontdict=d)
plt.legend(loc='upper left', fontsize=10)
plt.gca().set_facecolor('lightgray')
plt.gcf().set_facecolor('lightyellow')
for i,j in zip(teams, score):
    plt.text(i, j, j, ha='center', va='bottom', fontdict=d)
plt.tight_layout()
```

```python
score = [565, 688, 565, 953, 678]
teams = ['RR', 'KKR', 'MI', 'CSK', 'LSG']
d = { 'family': 'Times New Roman', 'color':  'black', 'weight': 'normal', 'size': 12}
plt.barh(teams, score, color=['red', 'green', 'blue', 'orange', 'purple'], height=0.8, label=teams )
plt.title('Highest Runs of Kohli in IPL', fontdict=d)
plt.xlabel('vs Teams', fontdict=d)
plt.ylabel('Runs Scored', fontdict=d)
plt.legend(loc='lower right', fontsize=10)
plt.gca().set_facecolor('lightgray')
plt.gcf().set_facecolor('lightyellow')
for i,j in zip(teams, score):
    plt.text(j, i, j, ha='left', va='center', fontdict=d)
plt.tight_layout()
```
